```python
import matplotlib.pyplot as plt

orders["order_purchase_timestamp"] = pd.to_datetime(orders["order_purchase_timestamp"])

monthly = orders.groupby(orders["order_purchase_timestamp"].dt.to_period("M")).size()
monthly.plot(kind="line", figsize=(10, 4), title="Orders per month")
plt.show()
```

```python
bad = orders[
    (orders["order_status"] == "delivered")
    & orders[["order_approved_at", "order_delivered_carrier_date", "order_delivered_customer_date"]].isna().any(axis=1)
]
print(len(bad))
```

```python
df = items.merge(orders[["order_id", "order_purchase_timestamp"]], on="order_id")

revenue = df.groupby(df["order_purchase_timestamp"].dt.to_period("M"))["price"].sum()
revenue = revenue["2017-01":"2018-08"]
revenue.plot(kind="line", figsize=(10, 4), title="Revenue per month")
plt.show()
```

```python
orders_per_person = customers.groupby("customer_unique_id").size()
print(orders_per_person.value_counts().head())
print((orders_per_person > 1).mean())
```

```python
multi = reviews[reviews["order_id"].duplicated(keep=False)].sort_values(["order_id", "review_creation_date"])
multi[["order_id", "review_id", "review_score", "review_creation_date"]].head(10)
```

```python
reviews_clean = (
    reviews.sort_values(["review_creation_date", "review_answer_timestamp"])
    .drop_duplicates("order_id", keep="last")
)
print(len(reviews_clean))
```

```python
print(reviews_clean["review_score"].value_counts().sort_index())
print(reviews_clean["review_score"].value_counts(normalize=True).sort_index().round(3))
```

```python
top = customers["customer_city"].value_counts().head(10)
top.sort_values().plot(kind="barh", figsize=(8, 4), title="Orders by city (top 10)")
plt.show()
```

```python
city_share = customers["customer_city"].value_counts(normalize=True).head(10) * 100

ax = city_share.sort_values().plot(kind="barh", figsize=(8, 4), title="Share of all orders by city (top 10)")
ax.set_xlabel("% of all orders")
ax.bar_label(ax.containers[0], fmt="%.1f%%", padding=3)
plt.show()
```

```python
tables = {
    "orders": orders, "items": items, "payments": payments,
    "reviews": reviews_clean, "customers": customers, "products": products,
    "sellers": sellers, "geo": geo, "translation": translation,
}

for name, df in tables.items():
    n = df.isna().sum()
    print(f"--- {name} ({len(df)} rows)")
    print(n[n > 0] if n.sum() > 0 else "no nulls")
```

```python
cols = ["product_category_name", "product_name_lenght", "product_description_lenght", "product_photos_qty"]
print(products[cols].isna().all(axis=1).sum())

dims = ["product_weight_g", "product_length_cm", "product_height_cm", "product_width_cm"]
print(products[dims].isna().all(axis=1).sum())
```

```python
no_cat = products[products["product_category_name"].isna()]["product_id"]
sold = items[items["product_id"].isin(no_cat)]

print(len(sold), len(items))
print(round(sold["price"].sum()), round(items["price"].sum()))
print(round(sold["price"].sum() / items["price"].sum() * 100, 2))
```
