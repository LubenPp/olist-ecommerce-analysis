```python
s = orders.merge(customers[["customer_id", "customer_state"]], on="customer_id")
s = s[s["order_status"] == "delivered"].dropna(subset=["order_delivered_customer_date"]).copy()
s["order_delivered_customer_date"] = pd.to_datetime(s["order_delivered_customer_date"])
s["order_estimated_delivery_date"] = pd.to_datetime(s["order_estimated_delivery_date"])
s = s[(s["order_purchase_timestamp"] >= "2017-01-01") & (s["order_purchase_timestamp"] < "2018-09-01")]

s["late"] = s["order_delivered_customer_date"] > s["order_estimated_delivery_date"]
s["days_early"] = (s["order_estimated_delivery_date"] - s["order_delivered_customer_date"]).dt.days

top5 = s["customer_state"].value_counts().head(5).index
g = (
    s[s["customer_state"].isin(top5)]
    .groupby("customer_state")
    .agg(orders=("order_id", "count"), late_share=("late", "mean"), median_days_early=("days_early", "median"))
)
g["late_share"] = g["late_share"] * 100
print(g.round(1))

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
g["late_share"].sort_values().plot(kind="barh", ax=axes[0], title="% of orders delivered late")
g["median_days_early"].sort_values().plot(kind="barh", ax=axes[1], title="Median days early vs estimate")
plt.tight_layout()
plt.show()
```

```python
s["diff_days"] = (s["order_delivered_customer_date"].dt.normalize() - s["order_estimated_delivery_date"].dt.normalize()).dt.days

top5 = s["customer_state"].value_counts().head(5).index
t = s[s["customer_state"].isin(top5)]

g2 = t.groupby("customer_state").agg(
    orders=("order_id", "count"),
    early_pct=("diff_days", lambda x: (x < 0).mean() * 100),
    late_pct=("diff_days", lambda x: (x > 0).mean() * 100),
    median_days_late=("diff_days", lambda x: x[x > 0].median()),
)
print(g2.round(1))
ax = g2["early_pct"].sort_values().plot(kind="barh", figsize=(8, 4), title="% of orders delivered before the promised date")
ax.set_xlabel("% early")
ax.bar_label(ax.containers[0], fmt="%.1f%%", padding=3)
plt.tight_layout()
plt.show()
ax = g2["median_days_late"].sort_values().plot(kind="barh", figsize=(8, 4), title="Median days late (late orders only)")
ax.set_xlabel("days late")
ax.bar_label(ax.containers[0], fmt="%.0f days", padding=3)
plt.tight_layout()
plt.show()
```

```python
o = (
    items.groupby("order_id").agg(price=("price", "sum"), freight=("freight_value", "sum"))
    .reset_index()
    .merge(orders[["order_id", "customer_id"]], on="order_id")
    .merge(customers[["customer_id", "customer_state"]], on="customer_id")
)
o["freight_pct_of_price"] = o["freight"] / o["price"] * 100

f = o.groupby("customer_state").agg(
    orders=("order_id", "count"),
    median_freight=("freight", "median"),
    median_freight_pct=("freight_pct_of_price", "median"),
)
print(f.sort_values("median_freight", ascending=False).round(1))

ax = f["median_freight"].sort_values().plot(kind="barh", figsize=(8, 7), title="Median freight per order, by state")
ax.set_xlabel("freight (R$)")
plt.tight_layout()
plt.show()
```

```python
r = s.merge(reviews_clean[["order_id", "review_score"]], on="order_id")
r["late"] = r["diff_days"] > 0
r["bad"] = r["review_score"] <= 2

print(len(r))
print(r.groupby("late")["bad"].agg(["mean", "count"]))
```

```python
r["bucket"] = pd.cut(r["diff_days"], bins=[-1000, -1, 0, 3, 7, 1000],
                     labels=["early", "on the day", "1-3 days late", "4-7 days late", "8+ days late"])
print(r.groupby("bucket", observed=True)["bad"].agg(["mean", "count"]))

res = r.groupby("bucket", observed=True)["bad"].mean() * 100
ax = res.plot(kind="bar", figsize=(8, 4), title="% bad reviews (1-2 stars) by delivery timing", rot=0)
ax.set_ylabel("% bad reviews")
ax.bar_label(ax.containers[0], fmt="%.1f%%", padding=3)
plt.tight_layout()
plt.show()
```

```python
top = r[r["customer_state"].isin(["SP", "RJ", "MG", "RS", "PR"])]
print(top.groupby("customer_state")["bad"].mean().round(3))
print(top.groupby(["customer_state", "late"])["bad"].mean().unstack().round(3))
```

## Key findings
1. Orders and revenue grew fast through 2017 and then plateaued in 2018. The data is reliable only for Jan 2017 - Aug 2018.
2. About 97% of customers ordered once. Only about 3% came back (rough figure: includes canceled orders, and recent customers had less time to return).
3. Orders delivered after the estimated date got a 1-2 star review 62% of the time, vs 9% for on-time orders. It rises with the delay: 32% (1-3 days late), 68% (4-7 days), 79% (8+ days).
4. Late orders are about 7% of orders but about a third of all bad reviews. The other two thirds come from on-time orders, so delivery is a big driver, not the only one.
5. RJ has the highest late rate among the 5 biggest states (12% vs 4-6%) and the most bad reviews (18% vs 11-12%). Its freight cost is typical, so shipping price doesn't explain it.

## Limits
- Correlation, not causation: a bad seller could cause both a late delivery and a bad review.
- "Late" is measured against Olist's own estimate.
- Only customers who left a review are included (latest review kept when there were several).
